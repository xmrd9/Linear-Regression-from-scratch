```python
#importing libraries
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class LinearReg:
    def __init__(self):
        pass
    def apply(self, X, Y):
        X0=np.array(X) #input values
        Y=np.array(Y)
        X=np.c_[np.ones((X0.shape[0],1)),X0] 
        beta=np.linalg.inv(X.T@X)@X.T@Y
        return X@beta
        
```

```python
df=pd.read_csv('medical_insurance.csv')
lin=LinearReg()
lin.apply(df.bmi, df.charges)
```
