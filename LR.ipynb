```python
class LinearRegression: # create linear regression class
    def __init__(self, w, b, m): # initialize w (weight), b(bias), and m (will be used for dataframe length)
        self.w=w
        self.b=b
        self.m=m
    def grad_desc(self, data, eta): #gradient descent
       w_grad=b_grad=0 
       for i in range(self.m):
           x=data.loc[i,'bmi'] # independent variable
           y=data.loc[i, 'charges'] # target variable
           w_grad += -2/self.m*x*(y-(self.w*x+self.b)) # partial derivative of mse with the respect to the weight
           b_grad += -2/self.m*(y-(self.w*x+self.b)) # partial derivative of mse with the respect to the bias
       self.w=self.w-eta*w_grad # update formulas
       self.b=self.b-eta*b_grad
              
```
